```python
print("\n" + "="*70)
print("DATA SPLIT (70-30 as per paper)")
print("="*70)

# Identify target column
target_col = 'sepsis3'

# Separate features and target
if target_col in df.columns:
    y = df[target_col]
    X_raw = df.drop(columns=[target_col])
else:
    raise ValueError(f"Target column '{target_col}' not found")

# Remove ID columns if present (they shouldn't be used for training)
id_cols = ['hadm_id', 'subject_id', 'icustay_id', 'row_id', 'patient_id']
for col in id_cols:
    if col in X_raw.columns:
        X_raw = X_raw.drop(columns=[col])
        print(f"Removed ID column: {col}")

# Split data: 70% training, 30% testing (as per paper)
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw, y, test_size=0.3, random_state=42, stratify=y
)

print(f"\nTraining set: {X_train_raw.shape[0]} samples")
print(f"Test set: {X_test_raw.shape[0]} samples")
print(f"\nTraining set target distribution:")
print(y_train.value_counts())
print(f"\nTest set target distribution:")
print(y_test.value_counts())
```

```python
class DataPreprocessor:
    """
    Data preprocessing following Jathanna et al. (2023)
    Fits on training data only, transforms both train and test
    """

    def __init__(self):
        self.scaler = MaxAbsScaler()
        self.selected_features = None # Will be set during fit
        self.feature_stats = None
        self.imputers = {}
        self.constant_features = []
        self.high_corr_features = []
        self.is_fitted = False

    def fit(self, X, y):
        """Fit preprocessing on training data only"""
        print("\n" + "="*70)
        print("FITTING PREPROCESSING ON TRAINING DATA")
        print("="*70)

        self.X_train = X.copy()
        self.y_train = y.copy()

        # Step 1: Remove constant features
        print("\n1. Identifying constant features...")
        self.constant_features = []
        for col in self.X_train.columns:
            if self.X_train[col].nunique() <= 1:
                self.constant_features.append(col)

        if self.constant_features:
            print(f"   Found {len(self.constant_features)} constant features")
        else:
            print("   No constant features found")

        # Get numerical columns after removing constant features
        numerical_cols = self.X_train.select_dtypes(include=[np.number]).columns.tolist()
        # Initialize selected_features with all numerical columns (before any further filtering)
        self.selected_features = [col for col in numerical_cols if col not in self.constant_features]
        print(f"   Initial numerical features after removing constant features: {len(self.selected_features)}")

        # Step 2: Impute missing values using iterative imputation (best as per paper)
        print("\n2. Fitting imputation on training data...")

        # Use iterative imputation with LightGBM (as per paper)
        self.iterative_imputer = IterativeImputer(
            estimator=LGBMRegressor(
                n_estimators=100,
                random_state=42,
                verbosity=-1
            ),
            max_iter=10,  # As per paper
            random_state=42,
            initial_strategy='median'
        )

        # Fit imputer on training data using the initially selected numerical columns
        self.iterative_imputer.fit(self.X_train[self.selected_features])
        print("   Iterative imputer fitted")

        # Step 3: Normalize using MaxAbsScaler (best as per paper)
        print("\n3. Fitting MaxAbsScaler on training data...")
        self.scaler.fit(self.X_train[self.selected_features])
        print("   MaxAbsScaler fitted")

        # Skipping Step 4: Feature selection using ANOVA F-test (as per user request)
        # Skipping Step 5: Remove highly correlated features (as per user request)

        print(f"   Final features to be used: {len(self.selected_features)}")

        self.is_fitted = True
        return self

    def transform(self, X):
        """Transform data using fitted preprocessing"""
        if not self.is_fitted:
            raise ValueError("Preprocessor must be fitted before transform")

        print("\nTransforming data...")

        # Ensure X has only the features we intend to transform
        X_to_transform = X[self.selected_features].copy()

        # Impute missing values
        X_imputed = pd.DataFrame(
            self.iterative_imputer.transform(X_to_transform),
            columns=self.selected_features,
            index=X.index
        )

        # Normalize
        X_normalized = pd.DataFrame(
            self.scaler.transform(X_imputed),
            columns=self.selected_features,
            index=X.index
        )

        # The final output is X_normalized, which already contains only self.selected_features
        X_final = X_normalized

        print(f"   Transformed shape: {X_final.shape}")
        print(f"   Features: {X_final.columns.tolist()}")

        return X_final

    def fit_transform(self, X, y):
        """Fit and transform training data"""
        self.fit(X, y)
        return self.transform(X)

# Initialize preprocessor
preprocessor = DataPreprocessor()

# Fit on training data and transform training data
X_train = preprocessor.fit_transform(X_train_raw, y_train)

# Transform test data
X_test = preprocessor.transform(X_test_raw)

print("\n" + "="*70)
print("PREPROCESSING COMPLETE")
print("="*70)
print(f"Training features: {X_train.shape[1]}")
print(f"Test features: {X_test.shape[1]}")
print(f"\nFeatures used: {X_train.columns.tolist()}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier, GradientBoostingClassifier, ExtraTreesClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from lightgbm import LGBMClassifier

def get_all_models():
    """
    Initialize all 12 models as per paper:
    1. Logistic Regression
    2. Naive Bayes
    3. Random Forest
    4. Decision Tree
    5. K-Nearest Neighbors
    6. SVM
    7. Linear Discriminant Analysis
    8. Quadratic Discriminant Analysis
    9. AdaBoost
    10. LightGBM
    11. Gradient Boosting
    12. Extra Trees
    """
    return {
        'LogisticRegression': LogisticRegression(random_state=42, max_iter=1000),
        'NaiveBayes': GaussianNB(),
        'RandomForest': RandomForestClassifier(random_state=42),
        'DecisionTree': DecisionTreeClassifier(random_state=42),
        'KNN': KNeighborsClassifier(),
        'SVM': SVC(probability=True, random_state=42),
        'LinearDiscriminant': LinearDiscriminantAnalysis(),
        'QuadraticDiscriminant': QuadraticDiscriminantAnalysis(),
        'AdaBoost': AdaBoostClassifier(random_state=42),
        'LightGBM': LGBMClassifier(random_state=42, verbosity=-1),
        'GradientBoosting': GradientBoostingClassifier(random_state=42),
        'ExtraTrees': ExtraTreesClassifier(random_state=42)
    }
```

```python
import optuna
from sklearn.model_selection import StratifiedKFold, cross_val_score

def optimize_hyperparameters(model_name, model, X_train, y_train, n_trials=30):
    """Optimize hyperparameters using Optuna"""
    print(f"\nOptimizing {model_name}...")

    def objective(trial):
        params = {}

        if model_name == 'AdaBoost':
            params = {
                'n_estimators': trial.suggest_int('n_estimators', 50, 500),
                'learning_rate': trial.suggest_float('learning_rate', 0.01, 1.0, log=True)
            }
        elif model_name == 'LightGBM':
            params = {
                'n_estimators': trial.suggest_int('n_estimators', 100, 1000),
                'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
                'num_leaves': trial.suggest_int('num_leaves', 10, 100),
                'min_child_samples': trial.suggest_int('min_child_samples', 10, 100)
            }
        elif model_name == 'RandomForest':
            params = {
                'n_estimators': trial.suggest_int('n_estimators', 100, 500),
                'max_depth': trial.suggest_int('max_depth', 5, 50),
                'min_samples_split': trial.suggest_int('min_samples_split', 2, 20),
                'min_samples_leaf': trial.suggest_int('min_samples_leaf', 1, 10),
                'max_features': trial.suggest_float('max_features', 0.1, 1.0)
            }
        elif model_name == 'GradientBoosting':
            params = {
                'n_estimators': trial.suggest_int('n_estimators', 100, 500),
                'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
                'max_depth': trial.suggest_int('max_depth', 3, 10),
                'min_samples_split': trial.suggest_int('min_samples_split', 2, 20)
            }
        elif model_name == 'ExtraTrees':
            params = {
                'n_estimators': trial.suggest_int('n_estimators', 100, 500),
                'max_depth': trial.suggest_int('max_depth', 5, 50),
                'min_samples_split': trial.suggest_int('min_samples_split', 2, 20),
                'max_features': trial.suggest_float('max_features', 0.1, 1.0)
            }
        else:
            return 0.5

        model.set_params(**params)

        cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
        scores = cross_val_score(model, X_train, y_train, cv=cv, scoring='roc_auc')

        return scores.mean()

    study = optuna.create_study(direction='maximize', study_name=f'{model_name}_optimization')
    study.optimize(objective, n_trials=n_trials, show_progress_bar=False)

    best_params = study.best_params
    model.set_params(**best_params)

    print(f"   Best AUROC: {study.best_value:.4f}")

    return model, best_params, study.best_value
```

```python
optimized_models = {}
best_params = {}
best_aurocs = {}

# Get all models
all_models = get_all_models()

# List of all 12 models to optimize (or at least initialize)
models_to_optimize = list(all_models.keys())

for model_name in models_to_optimize:
    if model_name in ['AdaBoost', 'LightGBM', 'RandomForest', 'GradientBoosting', 'ExtraTrees']:
        print(f"\n--- Optimizing {model_name} ---")
        model = all_models[model_name]
        optimized_model, params, auroc = optimize_hyperparameters(
            model_name, model, X_train_smote, y_train_smote, n_trials=30 # Reduced trials as per user request
        )
        optimized_models[model_name] = optimized_model
        best_params[model_name] = params
        best_aurocs[model_name] = auroc
    else:
        # For models not in the optimization list, use the default initialized model
        optimized_models[model_name] = all_models[model_name]
        # Evaluate these models with their default parameters to get a baseline AUROC
        cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
        try:
            scores = cross_val_score(all_models[model_name], X_train_smote, y_train_smote, cv=cv, scoring='roc_auc')
            default_auroc = scores.mean()
            best_aurocs[model_name] = default_auroc
            best_params[model_name] = {'note': 'Default parameters used', 'cv_auroc': default_auroc}
            print(f"Skipping hyperparameter optimization for {model_name}. Using default parameters. Default AUROC: {default_auroc:.4f}")
        except Exception as e:
            best_aurocs[model_name] = 0.0 # Placeholder in case of error
            best_params[model_name] = {'note': f'Skipped optimization and evaluation due to error: {e}'}
            print(f"Skipping {model_name} due to error during default evaluation: {e}")

print("\n" + "="*70)
print("HYPERPARAMETER OPTIMIZATION & BASELINE EVALUATION COMPLETE")
print("="*70)

print("\nResults for all Models:")
for name in models_to_optimize:
    print(f"- {name}: AUROC = {best_aurocs.get(name, 0.0):.4f}") # Use .get with default for safety
    print(f"  Best Params: {best_params.get(name, 'N/A')}")
```
